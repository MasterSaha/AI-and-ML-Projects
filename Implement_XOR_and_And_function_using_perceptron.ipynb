```python
import numpy as np
```

```python
def perceptron(x, w, b):
    net_input = np.dot(x, w) + b
    return 1 if net_input >= 0 else 0
```

```python
def AND_logic(x):
    w = np.array([1, 1])
    b = -1.5
    return perceptron(x, w, b)
```

```python
def OR_logic(x):
    w = np.array([1, 1])
    b = -0.5
    return perceptron(x, w, b)
```

```python
def NAND_logic(x):
    w = np.array([-1, -1])
    b = 2
    return perceptron(x, w, b)
```

```python
def XOR_logic(x):
    h1 = NAND_logic(x)
    h2 = OR_logic(x)
    final_input = np.array([h1, h2])
    return AND_logic(final_input)
```

```python
inputs = [
    np.array([0, 0]),
    np.array([0, 1]),
    np.array([1, 0]),
    np.array([1, 1])
]
print(inputs)
```

```python
print("AND Function")
for x in inputs:
    print(f"AND({x[0]}, {x[1]}) = {AND_logic(x)}")

print("\nXOR Function")
for x in inputs:
    print(f"XOR({x[0]}, {x[1]}) = {XOR_logic(x)}")
```
